## Exercise 3 - Draw Barchart




```python
categories = ['A', 'B', 'C', 'D']
values = [10, 15, 7, 12]

plt.title("Bar Chart Example")
plt.xlabel("Categories")
plt.ylabel("Values")
plt.bar(categories, values)
```

## Exercise 4 -  Draw Histogram


```python
import numpy as np
values = np.random.normal(50, 8, 200)

# Add the x- and y-axis labels and a title
plt.title("Histogram Example")
plt.xlabel("Value")
plt.ylabel("Frequency")

#Challenge: Change the number of bins to 20 and observe how the visual apperance of the histogram changes.
plt.hist(values, bins=20, color='blue', edgecolor='black')


```

## Exercise 5 -  Draw Piechart



```python
import pandas as pd

### cars dataset in the data folder
cars = pd.read_csv('../data/car-data.csv')

# Draw piechart for the 'Fuel_Type' column in the cars dataset
fuel_counts = cars['Fuel_Type'].value_counts()
plt.title("Pie Chart of Car Fuel Types")
plt.pie(fuel_counts, labels=fuel_counts.index, autopct='%1.1f%%', startangle=90)
plt.axis('equal')
plt.show()
```
