```python
'''
### In-Class Exercise 1: Create and Index an Array

- Import NumPy using the alias np.
- Create an array containing: 12, 24, 36, 48, 60.
- Print the complete array.
- Print the first element.
- Print the third element.
- Predict the output before running your code.
'''

import numpy as np

arr = np.array([12, 24, 36, 48, 60])
print("Complete array:", arr)
print("First element:", arr[0]) # Prediction: 12
print("Third element:", arr[2]) # Prediction: 36
```

```python
'''
### In-Class Exercise 2: Create a 2-D Array

- Create a 2-D array with two rows: [5, 10, 15] and [20, 25, 30].
- Print the complete array.
- Identify how many rows and values per row are visible in the array.
- Challenge: retrieve the value 25 using indexing.
'''

import numpy as np

arr = np.array([[5, 10, 15], [20, 25, 30]])
print("Complete array:\n", arr)
print("Shape:", arr.shape) # 2 rows and 3 columns
print("Value at [1, 1]:", arr[1, 1]) # 25
```

```python
'''
### In-Class Exercise 3: Slice an Array

- Create the array [2, 4, 6, 8, 10, 12, 14, 16].
- Slice the array so the result is [6, 8, 10, 12].
- Write down the starting and ending indexes you used.
- Explain why the ending index is not included in the result.
'''

import numpy as np

arr = np.array([2, 4, 6, 8, 10, 12, 14, 16])
result = arr[2:6]
print("Original array:", arr)
print("Sliced array:", result)
print("Starting index: 2")
print("Ending index: 6")
print("The ending index is not included in the result because array slicing in Python is exclusive of the end point.")
```

```python
'''
### In-Class Exercise 4: Sort Numeric and Text Data

- Sort the numeric array [45, 12, 78, 3, 30].
- Sort the text array ["orange", "apple", "banana", "kiwi"].
- Print both sorted arrays.
- Before running the code, predict the first and last values in each sorted result.
'''

import numpy as np

numbers = np.array([45, 12, 78, 3, 30])
words = np.array(["orange", "apple", "banana", "kiwi"])
print("Sorted numbers:", np.sort(numbers)) # predicted first value: 3, last value: 78
print("Sorted words:", np.sort(words)) # predicted first value: "apple", last value: "orange"
```

```python
### Exercise 1.5 Searching Arrays - find method of numpy to search an element in array - You can google for the method

# Find the indexes where the value is 9:


arr = np.array([3, 2, 14, 9, 10, 12])
indexes = np.where(arr == 9)
print("Indexes where value is 9:", indexes[0]) # Output: [3]
```
